# Experiment-3: Stemming, Lemmatization and Regular Expression

###Experiment 3.1:  Write a Python program to perform  Write a Python program to perform stemming using the Porter Stemmer. Apply stemming to the following words. Display the original words and their stems.
playing, played, plays, studies, studying, connected, connection, computers


```python
# Import the Porter Stemmer from NLTK
import nltk
from nltk.stem import PorterStemmer

# Ensure required resources are available
nltk.download('punkt')

# Initialize the stemmer
stemmer = PorterStemmer()

# List of words to stem as specified in the experiment
words = [
    "playing", "played", "plays",
    "studies", "studying",
    "connected", "connection",
    "computers"
]

# Print header
print(f"{'Original Word':<15} | {'Stemmed Word':<15}")
print("-" * 35)

# Perform stemming and display the results
for word in words:
    stemmed_word = stemmer.stem(word)
    print(f"{word:<15} | {stemmed_word:<15}")
```

###Experiment 3.2:  Write a Python program to perform lemmatization on a given input words:

cats,	dogs,   running,  runs,	ran, studies, studying,	better, children, mice,	went, ate,
leaves,	caring,


```python
import nltk
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')

# Input words
words = ["cats", "dogs", "running", "runs", "ran", "studies", "studying",
         "better", "children", "mice", "went", "ate", "leaves", "caring"]

# POS tags: n = noun, v = verb, a = adjective
pos = ['n', 'n', 'v', 'v', 'v', 'v', 'v', 'a', 'n', 'n', 'v', 'v', 'n', 'v']

# Create lemmatizer
lemmatizer = WordNetLemmatizer()

print(f"{'Original Word':<15} | {'Lemma':<15}")
print("-" * 35)
# Lemmatize each word according to its POS tag
for word, p in zip(words, pos):
    lemma = lemmatizer.lemmatize(word, p)
    print(f"{word:<15} | {lemma:<15}")
```

- Email addresses
- URLs
- Mobile numbers

```python
import re

text = """Welcome to the Natural Language Processing workshop! For more information,
visit https://www.nlpworkshop.com or www.python.org. You can contact the coordinator
at nlpworkshop@gmail.com or support@python.org. For registration, call +91-9876543210
or 9123456789. Follow us on social media @NLPWorkshop and @PythonLearner.
Share your experience using #NLP, #Python, and #MachineLearning.
You can also visit https://github.com/NLPWorkshop for the latest updates."""

# Find email addresses: username@domain.extension
emails = re.findall(r'\b[\w.-]+@[\w.-]+\.\w+\b', text)

# Find URLs starting with http/https or www
urls = re.findall(r'https?://\S+|www\.\S+', text)

# Find Indian mobile numbers with optional +91 country code
mobile_numbers = re.findall(r'(?:\+91[-\s]?)?[6-9]\d{9}\b', text)

# Find hashtags starting with #
hashtags = re.findall(r'#\w+', text)

# Find mentions starting with @
mentions = re.findall(r'@\w+', text)

print("EMAIL ADDRESSES:")
for item in emails:
    print(item)

print("\nURLS:")
for item in urls:
    print(item)

print("\nMOBILE NUMBERS:")
for item in mobile_numbers:
    print(item)

print("\nHASHTAGS:")
for item in hashtags:
    print(item)

print("\nMENTIONS:")
for item in mentions:
    print(item)
```
