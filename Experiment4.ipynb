### **4.1.** To perform Term-Frequency Analysis and Named Entity Recognition on a given text using any toolkit.
- Calculate the frequency of each word and save the output in CSV format (Term, Frequency)
- Identify the top 10 most frequent terms and display their corresponding frequencies.


```python
import re
from collections import Counter
import pandas as pd
import spacy

file_path = "/content/drive/MyDrive/CLNLP-LAB/Experiment 4/4.1_4.2_input.txt"

with open(file_path, "r", encoding="utf-8") as f:
    text = f.read()

# Term Frequency Analysis
words = re.findall(r"\b[a-zA-Z]+\b", text.lower())
word_counts = Counter(words)

# Save all term frequencies to CSV
df_freq = pd.DataFrame(
    word_counts.items(), columns=["Term", "Frequency"]
).sort_values(by="Frequency", ascending=False)

df_freq.to_csv("term_frequencies.csv", index=False)
print("Saved term frequencies to 'term_frequencies.csv'.\n")

# Display Top 10 Most Frequent Terms
print("--- Top 10 Most Frequent Terms ---")
for term, freq in word_counts.most_common(10):
    print(f"{term}: {freq}")

# Named Entity Recognition
nlp = spacy.load("en_core_web_sm")
doc = nlp(text)

print("\n--- Named Entities Identified ---")
for ent in doc.ents:
    print(f"{ent.text:<30} -> {ent.label_}")
```

### **4.2.** To perform Term-Frequency Analysis on a given text without using any toolkit.

- Calculate the frequency of each word and save the output in CSV format (Term, Frequency)
- Identify the top 10 most frequent terms and display their corresponding frequencies.


```python
import re

# Update this to your file's path in Google Drive
file_path = "/content/drive/MyDrive/CLNLP-LAB/Experiment 4/4.1_4.2_input.txt"

with open(file_path, "r", encoding="utf-8") as f:
    text = f.read()

# PART 1 & 2: Term Frequency Analysis Without Any Toolkit

# Tokenize and normalize words
words = re.findall(r"\b[a-zA-Z]+\b", text.lower())

# Calculate frequency using a dictionary
word_counts = {}

for word in words:
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

# Sort terms by frequency
sorted_words = sorted(
    word_counts.items(),
    key=lambda x: x[1],
    reverse=True
)

# 1. Save all term frequencies to a CSV file
with open("term_frequencies_without_toolkit.csv", "w", encoding="utf-8") as f:
    f.write("Term,Frequency\n")

    for term, frequency in sorted_words:
        f.write(f"{term},{frequency}\n")

print("Saved term frequencies to 'term_frequencies_without_toolkit.csv'.\n")

# 2. Display Top 10 Most Frequent Terms
print("--- Top 10 Most Frequent Terms ---")

for term, frequency in sorted_words[:10]:
    print(f"{term}: {frequency}")
```

### **4.3.** Write a Python program without using NLTK, spaCy, or any other NLP toolkit to:
1. Read multiple text documents.
2. Convert text to lowercase.
3. Remove punctuation.
4. Tokenize the documents.
5. Calculate TF for every term.
6. Calculate DF for every term.
7. Calculate IDF using the logarithmic formula.
8. Calculate TF-IDF for every term in every document.
9. Display Term, TF, DF, IDF, and TF-IDF at each step.
10. Identify the top 10 terms with the highest TF-IDF score for each document.


```python
import re
import math

# Input Documents
documents = [
    "Natural language processing is a field of artificial intelligence.",
    "Natural language processing helps computers understand human language.",
    "Machine learning is an important part of artificial intelligence."
]

# PART 1-4 : Preprocessing

processed_documents = []

for document in documents:                      # Read multiple text documents.
    document = document.lower()                 # Convert text to lowercase.
    document = re.sub(r"[^\w\s]", "", document) # Remove punctuation.
    tokens = document.split()                   # Tokenize the documents.
    processed_documents.append(tokens)

# Display Tokenized Documents
print("--- Tokenized Documents ---")

for i, tokens in enumerate(processed_documents, start=1):
    print(f"Document {i}: {tokens}")

# PART 5: Calculate TF

tf = []

for tokens in processed_documents:
    term_counts = {}

    for term in tokens:
        if term in term_counts:
            term_counts[term] += 1
        else:
            term_counts[term] = 1

    total_terms = len(tokens)

    document_tf = {}

    for term, count in term_counts.items():
        document_tf[term] = count / total_terms

    tf.append(document_tf)

# PART 6: Calculate DF

df = {}

for tokens in processed_documents:
    unique_terms = set(tokens)

    for term in unique_terms:
        if term in df:
            df[term] += 1
        else:
            df[term] = 1

# PART 7: Calculate IDF

total_documents = len(documents)

idf = {}

for term, frequency in df.items():
    idf[term] = math.log(total_documents / frequency)

# PART 8-9: Calculate TF-IDF and Display Results

for i in range(len(documents)):
    print(f"\n--- Document {i + 1} ---")
    print(f"{'Term':<15} {'TF':<10} {'DF':<5} {'IDF':<10} {'TF-IDF':<10}")

    tfidf_scores = {}

    for term in tf[i]:
        tfidf = tf[i][term] * idf[term]
        tfidf_scores[term] = tfidf

        print(
            f"{term:<15} "
            f"{tf[i][term]:<10.4f} "
            f"{df[term]:<5} "
            f"{idf[term]:<10.4f} "
            f"{tfidf:<10.4f}"
        )

    # PART 10: Top 10 Terms with Highest TF-IDF for each document.

    sorted_terms = sorted(
        tfidf_scores.items(),
        key=lambda x: x[1],
        reverse=True
    )

    print(f"\n--- Top 10 TF-IDF Terms for Document {i + 1} ---")

    for term, score in sorted_terms[:10]:
        print(f"{term}: {score:.4f}")
```
