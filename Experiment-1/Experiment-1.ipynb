# Experiment-1: Data Analysis

This notebook performs data analysis on **`employee_information_100.csv`** as per Experiment 1 requirements.
Designed for Google Colab and Jupyter Notebook environments.

### Setup & Data Loading

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set plot style
sns.set_theme(style="whitegrid")

# Load dataset (Ensure employee_information_100.csv is in your working directory / Colab session)
df = pd.read_csv('employee_information_100.csv')

print("Dataset loaded successfully! Shape:", df.shape)
df.head()
```

### Experiment 1.1: Read the employee dataset and display the average salary of each department using a bar chart.

```python
avg_salary_dept = df.groupby('Department')['Salary'].mean()
print("Average Salary by Department:")
print(avg_salary_dept.round(2))

plt.figure(figsize=(8, 5))
plt.bar(avg_salary_dept.index, avg_salary_dept.values, color='skyblue', edgecolor='black')
plt.title('Average Salary by Department', fontsize=14)
plt.xlabel('Department', fontsize=12)
plt.ylabel('Average Salary', fontsize=12)
plt.xticks(rotation=45)
plt.show()
```

### Experiment 1.2: Display the number of employees working in each department.

```python
emp_count_dept = df['Department'].value_counts()
print("Number of employees working in each department:")
print(emp_count_dept)

plt.figure(figsize=(8, 5))
plt.bar(emp_count_dept.index, emp_count_dept.values, color='coral', edgecolor='black')
plt.title('Number of Employees per Department', fontsize=14)
plt.xlabel('Department', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.xticks(rotation=45)
plt.show()
```

### Experiment 1.3: Display the percentage of male and female employees using a pie chart.

```python
gender_counts = df['Gender'].value_counts()
print("Gender Distribution:")
print(gender_counts)

plt.figure(figsize=(6, 6))
plt.pie(gender_counts, labels=gender_counts.index, autopct='%1.1f%%', colors=['#4C72B0', '#DD8452'], startangle=90, explode=(0.04, 0.04))
plt.title('Percentage of Male and Female Employees', fontsize=14)
plt.show()
```

### Experiment 1.4: Draw a histogram showing salary distribution.

```python
plt.figure(figsize=(8, 5))
plt.hist(df['Salary'], bins=10, color='teal', edgecolor='black', alpha=0.75)
plt.title('Salary Distribution Histogram', fontsize=14)
plt.xlabel('Salary', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.show()
```

### Experiment 1.5: Draw a scatter plot showing the relationship between experience and salary.

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['Experience_Years'], df['Salary'], color='purple', alpha=0.7, edgecolors='black', s=60)
plt.title('Relationship Between Experience and Salary', fontsize=14)
plt.xlabel('Experience (Years)', fontsize=12)
plt.ylabel('Salary', fontsize=12)
plt.show()
```

### Experiment 1.6: Display the top 10 employees with the highest salary.

```python
top_10_highest = df.nlargest(10, 'Salary')
print("Top 10 Employees with Highest Salary:")
display(top_10_highest)
```

### Experiment 1.7: Find the highest salary in every department.

```python
max_salary_dept = df.groupby('Department')['Salary'].max().reset_index()
max_salary_dept.columns = ['Department', 'Highest_Salary']
print("Highest Salary in Every Department:")
display(max_salary_dept)
```

### Experiment 1.8: Display employees whose salary is greater than the overall average salary.

```python
overall_avg = df['Salary'].mean()
print(f"Overall Average Salary: {overall_avg:.2f}\n")

above_avg_emp = df[df['Salary'] > overall_avg]
print(f"Total Employees Earning Above Average: {len(above_avg_emp)}")
display(above_avg_emp)
```

### Experiment 1.9: Calculate the average years of experience in each department.

```python
avg_exp_dept = df.groupby('Department')['Experience_Years'].mean().reset_index()
avg_exp_dept.columns = ['Department', 'Average_Experience_Years']
avg_exp_dept['Average_Experience_Years'] = avg_exp_dept['Average_Experience_Years'].round(2)
print("Average Years of Experience in Each Department:")
display(avg_exp_dept)
```

### Experiment 1.10: Display the distribution of employee ages using a histogram.

```python
plt.figure(figsize=(8, 5))
plt.hist(df['Age'], bins=10, color='salmon', edgecolor='black', alpha=0.75)
plt.title('Employee Age Distribution Histogram', fontsize=14)
plt.xlabel('Age', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.show()
```
