##5.1. Implement subword-level tokenization using a BPE-based tokenizer.Tokenize the sentence into subword units.Display the generated tokens and their corresponding token IDs.
(Implement two separate codes: a. Using pretrained model and b. Without using pretrained model)


###a. Using pretrained model

```python
# Import tokenizer
from transformers import AutoTokenizer

# Path of the uploaded file in Google Drive
file_path = "/content/drive/MyDrive/CLNLP-LAB/Experiment 5/input_sub_word_data.txt"

# Read the document
with open(file_path, "r", encoding="utf-8") as file:
    text = file.read()

# Load pretrained GPT-2 BPE tokenizer
tokenizer = AutoTokenizer.from_pretrained("gpt2")

# Tokenize the document
tokens = tokenizer.tokenize(text)

# Convert tokens into token IDs
token_ids = tokenizer.convert_tokens_to_ids(tokens)

# Display results
print("Tokens:")
print(tokens)

print("\nToken IDs:")
print(token_ids)
```

###b. Without using pretrained model

```python
# Import required libraries
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace

# Path of the uploaded file
file_path = "/content/drive/MyDrive/CLNLP-LAB/Experiment 5/input_sub_word_data.txt"

# Create a BPE tokenizer from scratch
tokenizer = Tokenizer(BPE())

# Split text into words before BPE tokenization
tokenizer.pre_tokenizer = Whitespace()

# Define the vocabulary size
trainer = BpeTrainer(
    vocab_size=1000,
    special_tokens=["[UNK]"]
)

# Train BPE tokenizer using the uploaded document
tokenizer.train([file_path], trainer)

# Read the same document
with open(file_path, "r", encoding="utf-8") as file:
    text = file.read()

# Tokenize the document
output = tokenizer.encode(text)

# Display tokens
print("Tokens:")
print(output.tokens)

# Display token IDs
print("\nToken IDs:")
print(output.ids)
```

 ## 5.2. Implement subword-level tokenization using a SentencePiece tokenizer. Given same input data of 5.1 Tokenize the sentence into subword units.
Display the generated tokens and their corresponding token IDs.

(Implement two separate codes: a. Using pretrained model and b.  Without using pretrained model)


###a. Using pretrained model

```python
# Import the pretrained tokenizer
from transformers import AutoTokenizer

# Path of the input file
file_path = "/content/drive/MyDrive/CLNLP-LAB/Experiment 5/input_sub_word_data.txt"

# Read the input document
with open(file_path, "r", encoding="utf-8") as file:
    text = file.read()

# Load pretrained T5 tokenizer
tokenizer = AutoTokenizer.from_pretrained("t5-small")

# Tokenize the document into subword units
tokens = tokenizer.tokenize(text)

# Convert tokens into token IDs
token_ids = tokenizer.convert_tokens_to_ids(tokens)

# Display tokens
print("Tokens:")
print(tokens)

# Display token IDs
print("\nToken IDs:")
print(token_ids)
```

###b. Without using pretrained model

```python
# Import SentencePiece
import sentencepiece as spm

# Path of the input file
file_path = "/content/drive/MyDrive/CLNLP-LAB/Experiment 5/input_sub_word_data.txt"

# Train SentencePiece tokenizer using the input document
spm.SentencePieceTrainer.train(
    input=file_path,
    model_prefix="sentencepiece_model",
    vocab_size=300,
    model_type="bpe"
)

# Load the trained tokenizer
tokenizer = spm.SentencePieceProcessor(
    model_file="sentencepiece_model.model"
)

# Read the input document
with open(file_path, "r", encoding="utf-8") as file:
    text = file.read()

# Tokenize the document into subword units
tokens = tokenizer.encode(text, out_type=str)

# Convert tokens into token IDs
token_ids = tokenizer.encode(text, out_type=int)

# Display tokens
print("Tokens:")
print(tokens)

# Display token IDs
print("\nToken IDs:")
print(token_ids)
```

##         5.3. Implement Part-of-Speech (POS) tagging on a given text and display the unique token, POS tag, and description of each POS tag.
"The young student is reading an interesting book in the library."

 (Implement two separate codes: a. Using Spacy and b.  NLTK )   


###a. Using spacy

```python
import spacy

nlp = spacy.load("en_core_web_sm")
text = "The young student is reading an interesting book in the library."

doc = nlp(text)

# Filter unique tokens while preserving POS and description
seen = set()
for token in doc:
    if token.text.lower() not in seen:
        seen.add(token.text.lower())
        print(f"{token.text:<12} {token.tag_:<6} {spacy.explain(token.tag_)}")
```

###a. Using nltk

```python
import nltk

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('averaged_perceptron_tagger', quiet=True)
nltk.download('averaged_perceptron_tagger_eng', quiet=True)
nltk.download('tagsets', quiet=True)

# Directly load the Penn Treebank tagset mapping from NLTK data
tagset = nltk.data.load('help/tagsets/upenn_tagset.pickle')

text = "The young student is reading an interesting book in the library."
tagged = nltk.pos_tag(nltk.word_tokenize(text))

seen = set()
for word, tag in tagged:
    if word.lower() not in seen:
        seen.add(word.lower())
        # tagset[tag][0] contains the concise description
        desc = tagset.get(tag, ('Punctuation / Symbol',))[0]
        print(f"{word:<12} {tag:<6} {desc}")
```

#       5.4. Implement Part-of-Speech (POS) tagging on a given text and display the unique token, POS tag, description of each token and frequency using Spacy.
    Input data: Same input data of 5.1


```python
import spacy
from collections import Counter

nlp = spacy.load("en_core_web_sm")

file_path = "/content/drive/MyDrive/CLNLP-LAB/Experiment 5/input_sub_word_data.txt"

with open(file_path, "r", encoding="utf-8") as file:
    text = file.read()

doc = nlp(text)

# Count frequencies of normalized tokens (excluding whitespace/newlines)
valid_tokens = [token.text.lower() for token in doc if not token.is_space]
counts = Counter(valid_tokens)

# Print unique token, POS tag, description, and frequency
seen = set()
print(f"{'Token':<15} {'POS':<6} {'Freq':<6} {'Description'}")
print("-" * 65)

for token in doc:
    key = token.text.lower()
    if not token.is_space and key not in seen:
        seen.add(key)
        desc = spacy.explain(token.tag_)
        freq = counts[key]
        print(f"{token.text:<15} {token.tag_:<6} {freq:<6} {desc}")
```
